```python
#Tupple are ordered collection of  verious type items that are immutible.
#creat empty tupple
tup=()
print(type(tup))
```

```python
#creat a tupple
fruits=('apple','banana','mango','cherry')
print(fruits)
```

```python
#creat mixed tupple
datas=("jayanta",21,78.65,True)
print(datas)
```

```python
#Acessing tupple element
for index,data in enumerate(datas):
    print(f"{index}.{data}")
print("print index 0 item:",datas[0])
print("print index 2 item:",datas[2])
print("print index 0-1 item:",datas[:2])
print("print all item:",datas[::])
print("print reverse item:",datas[::-1])
print("print index -3 item:",datas[-3])
```

```python
#tuple operation
concatinate_tupple=fruits+datas
print("List 1\n",fruits)
print("List 2\n",datas)
print("After concatinate two list\n",concatinate_tupple)
```

```python
print(datas)
print(datas.count("jayanta"))
print(datas.index("jayanta"))
```

```python
#pack tuple
pack_tupple="Name",23,3.4,True
print(pack_tupple)
print(type(pack_tupple))
```

```python
#unpack tupple
a,b,c,d=pack_tupple
print(a,b)
print(d)
```

```python
#unpacking with *
print(concatinate_tupple)
first,*middle,last=concatinate_tupple
print(first)
print (middle)
print(last)
```

```python
#nested tupple
nested_tupple=(('hello',3,True),(1,2,3,4),("hi","jayanta"))
print(nested_tupple[0])
print(nested_tupple[0][1:])
print(nested_tupple[1][2])
```

```python
for sub_tupple in nested_tupple:
    for item in sub_tupple:
        print(item,end=" ")
    print(" ")
```
