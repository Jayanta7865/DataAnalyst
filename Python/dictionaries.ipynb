```python
#Dictionaries are unordered collection of items,they  store data key value pairs,there key mast be unique and immutuble and value can be any type.
my_dict={'name':'jayanta','age':21,'college':"HIT"}
print(my_dict)
```

```python
#empty dict
empty_dict={}
print("This is a empty list:",empty_dict)
```

```python
#accessing dict element
print(my_dict)
for key,value in my_dict.items():
    print(key,value)
print(my_dict['name'])
print(my_dict['college'])
```

```python
#Acessing with get
print(my_dict.get('name'))
print(my_dict.get('roll'))
print(my_dict.get('roll',"not avilable")) #assing by defult value
```

```python
# before modify dict element
print(my_dict)
```

```python
#after modify
print(my_dict)
my_dict['name']='Sumana'
my_dict['age']=20
my_dict['address']="Medinipur"
print(my_dict)
```

```python
#delete key value pair
del my_dict['address']
print(my_dict)
```

```python
#method
keys=my_dict.keys()
print("All keys in my_dict:",keys)
values=my_dict.values()
print("All values in my_dict:",values)
items=my_dict.items()
print("All items in my_dict:",items)
```

```python
for key in my_dict:
    print(key)
for value in my_dict.values():
    print(value)   
for key,value in my_dict.items():
    print(key,value)
```

```python
my_dict_copy=my_dict
print(my_dict_copy)
print(my_dict)
```

```python
my_dict['name']="jayanta"
print(my_dict_copy)
print(my_dict)
```

```python
my_dict_copy1=my_dict.copy()
print(my_dict_copy1)
print(my_dict)
```

```python
my_dict["name"]="Sumana" #shallow copy
print(my_dict_copy1)
print(my_dict)
```

```python
#nested dictionaries
students={'student1':{"name":"jayanta","roll":20,"college":"HIT"},
         'student2':{"name":"Sumana","roll":21,"college":"HIHS"},
         'student3':{"name":"Mahadeb","roll":29,"college":"HIT"},
         'student4':{"name":"Ram","roll":28,"college":"HIM"}

}
print(students)
```

```python
for sub_student,item in students.items():
    print(sub_student,item)
```

```python
print(students['student1']['name'])
print(students['student2']['name'])
```

```python
for student in students.values():
    print(f"{student['name']},{student['roll']},{student['college']}")
```

```python
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
#dict comprehence
sqare={x:x**2 for x in range(5)}

print(sqare)
```

```python
#sqare of a odd number
squar={x:x**2 for x in range(11) if x%2==0}
print(squar)
```

```python
#user creat dist
my_dict = {}

try:
    n = int(input("Enter how many items in your dict: "))
except ValueError:
    print("Please enter a valid number!")
    n = 0

for i in range(n):
    key = input("Enter key: ")
    value = input("Enter value: ")
    my_dict[key] = value

print("\nFinal dictionary:")
print(my_dict)
```
