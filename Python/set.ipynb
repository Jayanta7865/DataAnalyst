```python
#creat a set
my_set={1,2,3,4,5,6}
print(my_set)
print(type(my_set))
```

```python
#creat empty set
empty_set=set()
print(empty_set)
```

```python
#creat set another way
new_set=set([1,2,3,4,5,6])
print(new_set)
```

```python
#add element in set
new_set.add(7)
print(new_set)
```

```python
#remove element in set
new_set.remove(6)
print(new_set)
```

```python
#remove element another way
new_set.discard(7)
print(new_set)
```

```python
#pop element in set
#set pop first element
pop_item=new_set.pop()
print(pop_item)
print(new_set)
```

```python
#clear whole element in set
new_set.clear()
print(new_set)
```

```python
#set can't contain duplicate values
set1={1,2,3,2,3,4,5,7,10,7}
print(set1)
```

```python
#set operation
#union
num_set1={1,2,3,4,5,6,7}
num_set2={5,6,7,8,9,10}
union_set=num_set1.union(num_set2)
print(union_set)

```

```python
#intersection (can't update original set it creat a new set)
inter_set=num_set1.intersection(num_set2)
print(inter_set)
print(num_set1)
```

```python
#intersection_update (update originalset)
set1={1,2,3,4,5,6,7}
set2={5,6,7,8,9,10}
print("set1 before update",set1)
set1.intersection_update(set2)
print("After update",set1)
```

```python
#difference (print difference between two set)
set1={1,2,3,4,5,6,7}
set2={5,6,7,8,9,10}
print("print set1 Elements difference from set2",set1.difference(set2))
print("print set2 Elements difference from set1",set2.difference(set1))
```

```python
#difference update
print("set1 before update",set1)
set1.difference_update(set2)
print("After update",set1)
```

```python
#symmetric difference
set1={1,2,3,4,5,6,7}
set2={5,6,7,8,9,10}
print(set1.symmetric_difference(set2))
```

```python
#issubset,issuperset
set1={5,6,7}
set2={5,6,7,8,9,10}
print(set1.issubset(set2))
print(set2.issuperset(set1))
```
